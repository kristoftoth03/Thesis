```python
def thermal_conductivity(T0, Rho0):

    #Lambda0 meghatározása
    Tr = T0 / Tc
    Delta = Rho0 / Rhoc

    Lambda0_sum = 0
    i = 0
    for i in range(0,4):
        Lambda0_sum += Lk[i] / Tr**i
        #print(i)
        #print(Lk[i])

    Lambda0 = np.sqrt(Tr) / (Lambda0_sum) * 0.001
    #print(Lambda0)

    Lambda_res = 0
    i = 0
    #range 0-6, kitevobe i+1 ??
    for i in range(0,6):
        Lambda_res += (B1[i] + B2[i] * Tr) * Delta**(i +1)
        #print(B1[i])
        #print(i+1)
    #print(Lambda_res)

    Lamdba_c = 0.001*(-17.47-44.88*(T0/Tc -1)) / (0.8563-np.exp(8.865*(T0/Tc -1) + 4.16*(Rho0/Rhoc -1)**2 + 2.303*(T0/Tc -1)*(Rho0/Rhoc -1)- (Rho0/Rhoc -1)**3) - 0.4503*(Rho0/Rhoc -1) - 7.197*(T0/Tc -1))
    #print(Lamdba_c)

    Lambda = Lambda0 + Lambda_res + Lamdba_c

    return Lambda

    #print('Lambda : ' + str(Lambda))
```

```python

# Initialize storage lists
#beta_p0_list = []
#chi_T0_list = []
t_heating_list = []
Delta_t_list = []
num_steps_list = []
tau_diff_list = []
tau_acustic_list = []


# Initialize results storage for temperature and pressure
T_all_cases = []
p_all_cases = []
rho_all_cases = []

#added these 2025.09.17.
cp_all_cases = []
cv_all_cases = []
lambda_all_cases = []
muJT_all_cases = []
as_all_cases = []
betap_all_cases = []
khit_all_cases = []

# Constants
Nx = 20#100    
Delta_x = Lh / Nx

for i in range(1): #range(5)-t holnap kicserélem range 1-ra hogy lefusson legalább egy eset és látszódjon valami
    print(f"Running case {i+1}/5...")

    T_i = T0ic[i] # IF ISOTHERMAL THIS IS 0, IF ISOBARIC THIS IS i----------------------------------------------------------------------------------------------------------------------------------------------------
    rho_i = np.full(Nx, rho0[i])
    cp_i  = np.full(Nx, cp0[i])
    cv_i  = np.full(Nx, cv0[i])
    lambda_i = np.full(Nx, lambda0[i])
    mu_JT_i = np.full(Nx, mu_JT0[i])
    as_i = np.full(Nx, as0[i])

    beta_p = np.zeros(Nx)
    chi_T  = np.zeros(Nx)

    for n in range(0,Nx):
        #v0 = 1 / rho_i
        #beta_p0 = (cp0_i * mu_JT_i + v0) / (v0 * T0ic_i)
        #Beta_p_result[j,i] = ( mu_JT_result[j,i] * cp_result[j,i] * rho_result[j,i] + 1 ) / T[i]
        beta_p[n] = (mu_JT_i[n] * cp_i[n] * rho_i[n] + 1) / T_i #itt kiszámolom a fenti alapján, ezt használtam eddig

        #chi_T0 = (T0ic_i * v0 * beta_p0**2 / cv0_i) / (cp0_i / cv0_i - 1)
        #Khi_T_result[j,i] = T[i] * (1 / rho_result[j,i]) * Beta_p_result[j,i]**2 / ( cp_result[j,i] - cv_result[j,i] )  tehát ugyan azokat a képletek használtam 

        #chi_T[n] = T_i * (1 / rho_i[n]) * beta_p[n]**2 / (cp_i[n] - cv_i[n]) #itt is az eddigi diagramos keplet alapjan szmolok
        
        #Khi_T -t hangsebességgel számolva: 2025.09.20.
        chi_T[n]= cp_i[n] / ( cv_i[n] * rho_i[n] * as_i[n]**2 )



    tau_diff = (Lh**2 * rho0[i] * cp0[i]) / lambda0[i] # itt kicseréltem a list rho_i .... -okat az eredeti kezdeti értékre
    Delta_t = min(Delta_x**2 / (2*lambda0[i]/(rho0[i]*cp0[i])), t_heating / 100)  #<-------
                                                                    #2025.12.13. Ezt átírtam 100 ra mert 10 mellett th=1000 mellett rho nem
                                                                    #konergált, valószínűleg azért mert túl nagy volt dt

    num_steps = int(tau_diff / Delta_t ) #----------------------------------------------2025.10.24. EZT átírtam /3, mert különben 5x annyit számolna
                                                                                                        #tehát kvízi a diffóziós idő harmadáig fut most  
    tau_acustic = Lh / as0[i]

    # Store  parameters
    tau_diff_list.append(tau_diff)
    t_heating_list.append(t_heating)
    tau_acustic_list.append(tau_acustic)
    Delta_t_list.append(Delta_t)
    num_steps_list.append(num_steps)
    #beta_p0_list.append(beta_p)
    #chi_T0_list.append(chi_T)



    # --- INITIALIZATION ---
    T_prev = np.zeros((num_steps+1, Nx))    # T_star[j, n] = T at time j, position n
    q_prev = np.zeros((num_steps+1, Nx+1))  # q_star[j, n] = q at time j, position n
    p_prev = np.zeros(num_steps+1)          # p_star[j] = p at time j
    rho_save = np.zeros((num_steps+1, Nx))

    #added these 2025.09.17.
    cp_save = np.zeros((num_steps+1, Nx))
    cv_save= np.zeros((num_steps+1, Nx))
    lambda_save = np.zeros((num_steps+1, Nx))
    muJT_save = np.zeros((num_steps+1, Nx))
    as_save = np.zeros((num_steps+1, Nx))
    betap_save = np.zeros((num_steps+1, Nx))
    khit_save = np.zeros((num_steps+1, Nx))


    # Initial conditions
    T_prev[0, :] = T_i                # Initial dimensionless temperature profile
    p_prev[0] = p0[0]                     #  IF ISOTHERMAL THIS IS i, IF ISOBARIC THIS IS 0-------------------------------------------------------------------------------------------------------------------------------


    rho_save[0,:] = rho_i

    #added these 2025.09.17.
    cp_save[0,:] = cp_i
    cv_save[0,:]= cv_i
    lambda_save[0,:] = lambda_i
    muJT_save[0,:] = mu_JT_i
    as_save[0,:] = as_i
    betap_save[0,:] = beta_p
    khit_save[0,:] = chi_T

    tolerance = 1e-9                    # Set your desired convergence limit

    # --- TIME STEPPING LOOP ---
    T_next = np.zeros(Nx)
    p_next = 0 
    p_init = 0
    q_next = np.zeros(Nx+1)

    for j in tqdm(range(num_steps)):
        if j * Delta_t < t_heating: # SWITCHED <= to <          ---------2025.8.27.
            q_prev[j, 0] = q0
        else:
            q_prev[j, 0] = 0 

        # 1. coupled temperature and pressure update
        k=0
        p_next = 0


        while abs(p_next - p_init) >= tolerance or k == 0:
            dp = 0
            numerator = 0
            denominator = 0

            if k == 0:
                for n in range(0,Nx):
                    T_next[n] = (
                        T_prev[j,n]
                        + Delta_t *(1 / (rho_i[n]*cp_i[n])) * (q_prev[j, n] - q_prev[j, n+1]) / Delta_x
                    )

                k+=1

                # 2. Pressure update     
                for n in range(0,Nx): 

                    numerator = numerator + rho_i[n] * beta_p[n] * (T_next[n]-T_prev[j, n]) /Delta_t * Delta_x 

                    denominator = denominator + rho_i[n] * chi_T[n]  * Delta_x

                dp = numerator / denominator * Delta_t
                p_init = p_prev[j] + dp


            else: #mostmár k nem egyenlő 0, számítok egy új hőmérsékletet a nyomással kapcsolva
                k+=1
                dp = 0
                numerator = 0
                denominator = 0

                p_init = p_next if k > 2 else p_init

                for n in range(0,Nx):
                    T_next[n] = (
                        T_prev[j,n]
                        + Delta_t * (1 / (rho_i[n]*cp_i[n])) * ((q_prev[j, n] - q_prev[j, n+1]) / Delta_x
                        +  T_prev[j,n] * beta_p[n] * (p_init - p_prev[j]) / Delta_t)) #újraszámítom a k=0 nál számolt T_next et, az ott kapott p figyelembevételével 
                         
                # 2. Pressure update     
                for n in range(0,Nx): 
                    numerator = numerator + rho_i[n] * beta_p[n] * (T_next[n]-T_prev[j, n]) / Delta_t* Delta_x 

                    denominator = denominator + rho_i[n] * chi_T[n] * Delta_x

                dp = numerator / denominator * Delta_t
                p_next = p_prev[j] + dp #számítok egy új nyomást
                
                
            
        #print(T_next[0])
        #itt bekonvergált, úgyhogy itt számítom ki az anyagi paramétereket az új hőmérsékleten és nyomáson
        # ebben a szekcióban fogom kiszámolni ugyan azokat az adatokat az állapotegyenlettel 
        
        for n in range(0,Nx):
            rho_guess = rho_i[n]
            try:
                rho_i[n] = newton(
                density_from_pressure,
                rho_guess,
                args=(T_next[n], p_next),
                tol=1e-1, #e-9
                maxiter=50000
             )
            except RuntimeError:
                # Fallback: if Newton fails, try a small perturbation
                rho_i[n] = newton(
                density_from_pressure,
                rho_guess * 1.05,
                args=(T_next[n], p_next),
                tol=1e-1,
                maxiter=50000
             )

            lambda_i[n] = thermal_conductivity(T_next[n],rho_i[n])

            cp_i[n] = get_cp(T_next[n],rho_i[n])

            cv_i[n] = get_cv(T_next[n],rho_i[n])

            as_i[n] = speed_of_sound(T_next[n],rho_i[n])

            mu_JT_i[n] = joule_thompson(T_next[n],rho_i[n])

            for n in range(0,Nx):
                beta_p[n] = (mu_JT_i[n] * cp_i[n] * rho_i[n] + 1) / (T_next[n]) #itt kiszámolom a fenti alapján, ezt használtam eddig

                chi_T[n] =  cp_i[n] / ( cv_i[n] * rho_i[n] * as_i[n]**2 ) #itt is az eddigi diagramos keplet alapjan szmolok


        # 3. Heat flux update (Fourier law)
        for n in range(1, Nx):
            q_next[n] = (T_next[n-1] - T_next[n]) / Delta_x * lambda_i[n]


        # 4. Advance to next time step
        T_prev[j+1, :] = T_next
        q_prev[j+1, :] = q_next
        p_prev[j+1] = p_next
        rho_save[j+1, :] = rho_i

        cp_save[j+1,:] = cp_i
        cv_save[j+1,:]= cv_i
        lambda_save[j+1,:] = lambda_i
        muJT_save[j+1,:] = mu_JT_i
        as_save[j+1,:] = as_i
        betap_save[j+1,:] = beta_p
        khit_save[j+1,:] = chi_T



    # Save results for plotting
    T_all_cases.append(T_prev)
    p_all_cases.append(p_prev)
    rho_all_cases.append(rho_save)

    #added these 2025.09.17.
    cp_all_cases.append(cp_save)
    cv_all_cases.append(cv_save)
    lambda_all_cases.append(lambda_save)
    muJT_all_cases.append(muJT_save)
    as_all_cases.append(as_save)
    betap_all_cases.append(betap_save)
    khit_all_cases.append(khit_save)



    #as of 2025.09.17. im saving all material properties since restoring them takes as long as the simulation itself
    #also keep in mind that the first simulation ran with e-1 tolarence for rho

```
